##**Conhecendo os dados**

### Importando os dados

```python
import pandas as pd
```

```python
df = pd.read_csv('imigrantes_canada.csv')
df.info()
```

### Realizando alguns tratamentos

```python
df.set_index('País', inplace=True)
anos = list(map(str, range(1980, 2014)))
anos
```

### Extraindo a série de dados para o Brasil

```python
brasil = df.loc['Brasil', anos]
brasil_dict = {'ano': brasil.index.tolist(), 'imigrantes': brasil.values.tolist()}

dados_brasil = pd.DataFrame(brasil_dict)
dados_brasil
```

```python
import matplotlib.pyplot as plt
```

## Evolução de imigrantes brasileiros para o Canadá

```python
fig, ax = plt.subplots()
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Imigração do Brasil para o Canadá\n1980 a 2013', fontsize=18, loc='left')
ax.set_xlabel('Ano', fontsize=14)
ax.set_ylabel('Número de imigrantes', fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import os

# Dados de exemplo (você pode usar seus próprios dados)
dados_exemplo = {
    'Pais': ['Brasil', 'Argentina', 'Chile', 'Colombia', 'Peru'],
    'Imigrantes_Canada': [15000, 12000, 10000, 8000, 7000]
}
df_exemplo = pd.DataFrame(dados_exemplo)

# Cria o diretório 'imagens' se ele não existir
if not os.path.exists('imagens'):
    os.makedirs('imagens')

# Cria o gráfico
fig, ax = plt.subplots(figsize=(10, 6)) # Ajusta o tamanho da figura

ax.bar(df_exemplo['Pais'], df_exemplo['Imigrantes_Canada'], color='skyblue')
ax.set_xlabel('País', fontsize=14)
ax.set_ylabel('Número de Imigrantes para o Canadá', fontsize=14)
ax.set_title('Imigração de Países da América do Sul para o Canadá', fontsize=16)
ax.tick_params(axis='x', rotation=45) # Rotaciona os rótulos do eixo X para melhor visualização
plt.tight_layout() # Ajusta o layout para evitar sobreposição

# Salva o gráfico no diretório 'imagens'
plt.savefig('imagens/grafico_imigracao_america_do_sul.png')

plt.show()
```
